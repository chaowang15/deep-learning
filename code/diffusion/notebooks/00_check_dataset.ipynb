```python
import matplotlib.pyplot as plt
import torch
from torchvision.utils import make_grid

from src.datasets import DatasetConfig, denormalize, image_settings_for_dataset, make_dataloader

print("torch:", torch.__version__)
print("cuda:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("gpu:", torch.cuda.get_device_name(0))
```

## MNIST

Expected shape: `[batch, 1, 28, 28]`. Pixel values should be normalized to roughly `[-1, 1]`.

```python
image_size, channels = image_settings_for_dataset("mnist")
mnist_cfg = DatasetConfig(name="mnist", image_size=image_size, channels=channels, batch_size=32, num_workers=0)
mnist_loader = make_dataloader(mnist_cfg, train=True)
mnist_images, mnist_labels = next(iter(mnist_loader))

print("images:", mnist_images.shape)
print("labels:", mnist_labels[:16].tolist())
print("min/max:", mnist_images.min().item(), mnist_images.max().item())
```

```python
grid = make_grid(denormalize(mnist_images[:32]), nrow=8)
plt.figure(figsize=(6, 4))
plt.imshow(grid.permute(1, 2, 0), cmap="gray")
plt.axis("off");
```
